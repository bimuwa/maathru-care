```python
# Start again from the cleaned Book2 dataset
df = book2_clean.copy()

# Drop unsuitable columns by their positions
columns_to_drop_by_position = [0, 8, 9, 11, 13]

df_model = df.drop(
    columns=df.columns[columns_to_drop_by_position]
).copy()

# Rename the 13 remaining columns
df_model.columns = [
    "Age",
    "Gravida",
    "TetanusVaccination",
    "GestationalAge",
    "WeightKg",
    "HeightFt",
    "BloodPressure",
    "FetalPosition",
    "FetalHeartRate",
    "UrineSugar",
    "VDRL",
    "HBsAg",
    "HighRiskPregnancy"
]

print("Shape:", df_model.shape)
print("\nColumns:")
print(df_model.columns.tolist())
```

```python
# Convert Age to numeric
df_model["Age"] = pd.to_numeric(df_model["Age"], errors="coerce")

# Extract the number from values like "38 week"
df_model["GestationalAge"] = (
    df_model["GestationalAge"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
    .astype(float)
)

# Extract the number from values like "50 kg"
df_model["WeightKg"] = (
    df_model["WeightKg"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
    .astype(float)
)

# Extract the number from values like 5.3"
df_model["HeightFt"] = (
    df_model["HeightFt"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
    .astype(float)
)

# Extract the number from values like "140m"
df_model["FetalHeartRate"] = (
    df_model["FetalHeartRate"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
    .astype(float)
)

print(df_model[
    ["Age", "GestationalAge", "WeightKg", "HeightFt", "FetalHeartRate"]
].head())
```

```python
# Split blood pressure into systolic and diastolic values
bp_split = df_model["BloodPressure"].astype(str).str.split("/", expand=True)

df_model["SystolicBP"] = pd.to_numeric(bp_split[0], errors="coerce")
df_model["DiastolicBP"] = pd.to_numeric(bp_split[1], errors="coerce")

# Remove the original text blood pressure column
df_model = df_model.drop(columns=["BloodPressure"])

print(df_model[["SystolicBP", "DiastolicBP"]].head())
```

```python
# Convert binary categorical columns

binary_mappings = {
    "TetanusVaccination": {
        "Yes": 1,
        "No": 0
    },
    "UrineSugar": {
        "Yes": 1,
        "No": 0
    },
    "VDRL": {
        "Positive": 1,
        "Negative": 0
    },
    "HBsAg": {
        "Positive": 1,
        "Negative": 0
    }
}

for column, mapping in binary_mappings.items():
    df_model[column] = (
        df_model[column]
        .astype(str)
        .str.strip()
        .map(mapping)
    )

print(
    df_model[
        ["TetanusVaccination", "UrineSugar", "VDRL", "HBsAg"]
    ].head()
)
```

```python
# Convert Gravida
df_model["Gravida"] = (
    df_model["Gravida"]
    .astype(str)
    .str.extract(r"(\d+)")[0]
    .astype(float)
)

print(df_model["Gravida"].head())
```

```python
# Convert Fetal Position
df_model["FetalPosition"] = (
    df_model["FetalPosition"]
    .astype(str)
    .str.strip()
    .map({
        "Normal": 0,
        "Abnormal": 1
    })
)

print(df_model["FetalPosition"].value_counts(dropna=False))
```

```python
# Convert target variable
df_model["HighRiskPregnancy"] = (
    df_model["HighRiskPregnancy"]
    .astype(str)
    .str.strip()
    .map({
        "Yes": 1,
        "No": 0
    })
)

print(df_model["HighRiskPregnancy"].value_counts(dropna=False))
```

```python
# Step 8: Separate features (X) and target (y)

X = df_model.drop(columns=["HighRiskPregnancy"])
y = df_model["HighRiskPregnancy"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Create the baseline model
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        random_state=42,
        max_iter=1000
    ))
])

# Train using ONLY the training data
logistic_model.fit(X_train, y_train)

print("Logistic Regression training completed successfully.")
```

```python
# Predict the unseen test data
y_pred_logistic = logistic_model.predict(X_test)

print("Predictions generated:", len(y_pred_logistic))
```

```python
from sklearn.metrics import accuracy_score

accuracy_logistic = accuracy_score(y_test, y_pred_logistic)

print(f"Logistic Regression Accuracy: {accuracy_logistic * 100:.2f}%")
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_logistic,
        target_names=["Not High Risk", "High Risk"]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic))
```

```python
print("===== BASELINE RESULT =====")
print(f"Model: Logistic Regression")
print(f"Accuracy: {accuracy_logistic * 100:.2f}%")
```

```python
# Step 24F: Calculate mean absolute SHAP importance

shap_importance = pd.DataFrame({
    "Feature": X_test.columns,
    "MeanAbsSHAP": abs(shap_values).mean(axis=0)
}).sort_values(
    by="MeanAbsSHAP",
    ascending=False
).reset_index(drop=True)

print("===== SHAP FEATURE IMPORTANCE =====")
print(shap_importance)

print("\n===== TOP 5 FEATURES =====")
print(shap_importance.head(5))
```

```python
# Step 24G: Compare Random Forest and SHAP feature importance

importance_comparison = feature_importance.copy()

importance_comparison = importance_comparison.rename(
    columns={"Importance": "RandomForestImportance"}
)

importance_comparison = importance_comparison.merge(
    shap_importance,
    on="Feature",
    how="left"
)

importance_comparison = importance_comparison.sort_values(
    by="MeanAbsSHAP",
    ascending=False
).reset_index(drop=True)

print("===== RANDOM FOREST vs SHAP IMPORTANCE =====")
print(importance_comparison)
```

```python
# Step 24H: Measure agreement between Random Forest and SHAP rankings

from scipy.stats import spearmanr

rank_data = importance_comparison.copy()

# Create ranks
rank_data["RF_Rank"] = (
    rank_data["RandomForestImportance"]
    .rank(ascending=False, method="average")
)

rank_data["SHAP_Rank"] = (
    rank_data["MeanAbsSHAP"]
    .rank(ascending=False, method="average")
)

# Calculate Spearman rank correlation
correlation, p_value = spearmanr(
    rank_data["RF_Rank"],
    rank_data["SHAP_Rank"]
)

print("===== FEATURE IMPORTANCE RANK AGREEMENT =====")
print(f"Spearman Rank Correlation: {correlation:.4f}")
print(f"P-value: {p_value:.6f}")

print("\nFeature Ranking Comparison:")
print(
    rank_data[
        [
            "Feature",
            "RF_Rank",
            "SHAP_Rank",
            "RandomForestImportance",
            "MeanAbsSHAP"
        ]
    ].sort_values("SHAP_Rank")
)
```

```python
# Step 24I: Visual comparison of Random Forest and SHAP rankings

import matplotlib.pyplot as plt
import numpy as np

plot_data = importance_comparison.copy()

# Rank features according to SHAP importance
plot_data = plot_data.sort_values(
    by="MeanAbsSHAP",
    ascending=True
)

y_pos = np.arange(len(plot_data))

plt.figure(figsize=(10, 7))

plt.barh(
    y_pos - 0.18,
    plot_data["RandomForestImportance"],
    height=0.35,
    label="Random Forest"
)

# Normalize SHAP values only for visual comparison
shap_normalized = (
    plot_data["MeanAbsSHAP"] /
    plot_data["MeanAbsSHAP"].max() *
    plot_data["RandomForestImportance"].max()
)

plt.barh(
    y_pos + 0.18,
    shap_normalized,
    height=0.35,
    label="SHAP (normalized)"
)

plt.yticks(y_pos, plot_data["Feature"])
plt.xlabel("Relative Importance")
plt.ylabel("Feature")
plt.title("Random Forest vs SHAP Feature Importance")
plt.legend()
plt.tight_layout()
plt.show()
```
